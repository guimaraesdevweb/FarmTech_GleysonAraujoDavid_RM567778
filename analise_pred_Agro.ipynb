```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler

print("Bibliotecas importadas com sucesso!")
```

```python
try:
    dados = pd.read_csv('dados.csv')
    display(dados.head()) # O 'display' no Jupyter renderiza uma tabela elegante
except FileNotFoundError:
    print("Erro: O arquivo 'dados.csv' não foi encontrado.")
```

```python
# 2. SEPARAR VARIÁVEIS
X = dados[['precipitacao', 'umidade_especifica', 'umidade_relativa', 'temperatura']]
y = dados['rendimento']

# 3. PADRONIZAR
scaler = StandardScaler()
X_padronizado = scaler.fit_transform(X)

# 4. DIVIDIR EM TREINO E TESTE
X_train, X_test, y_train, y_test = train_test_split(X_padronizado, y, test_size=0.2, random_state=42)

print("Dados padronizados e divididos!")
```

```python
modelos = {
    'RLM': LinearRegression(),
    'AD': DecisionTreeRegressor(),
    'RF': RandomForestRegressor(random_state=42),
    'SVM': SVR(),
    'RNA': MLPRegressor(max_iter=5000, random_state=42)
}

resultados = []

for nome, modelo in modelos.items():
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)
    
    resultados.append({
        'Modelo': nome,
        'R2': r2_score(y_test, y_pred),
        'MSE': mean_squared_error(y_test, y_pred),
        'MAE': mean_absolute_error(y_test, y_pred)
    })

# Criando o DataFrame
df_resultados = pd.DataFrame(resultados).sort_values(by='R2', ascending=False)
display(df_resultados)
```

```python
# Usando o RLM como exemplo
rlm_model = modelos['RLM']
dados['Previsao_Modelo'] = rlm_model.predict(X_padronizado)

display(dados[['rendimento', 'Previsao_Modelo']].head(10))
```
